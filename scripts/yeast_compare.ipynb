```python
# Normalize each pair so that (A, B) and (B, A) are treated as the same
# We'll create a sorted tuple for each pair
def normalize_pair(row):
    return tuple(sorted([row['IDR1'], row['IDR2']]))

new_df['normalized_pair'] = new_df.apply(normalize_pair, axis=1)

print("\nDataFrame with normalized pairs:")
display(new_df)

# Get the unique normalized pairs
unique_matches = new_df['normalized_pair'].unique().tolist()

print("\nUnique bidirectional matches (normalized):")
print(unique_matches)

# rows corresponding to these unique matches
unique_rows_df = new_df.drop_duplicates(subset=['normalized_pair']).drop(columns=['normalized_pair'])

print("\nOriginal rows representing unique bidirectional matches:")
display(unique_rows_df)
```
