```python
# Assuming your FASTA file is named 'example.fasta' and is in the content directory
fasta_file = "/home/wenyuantong/Desktop/data/PsyTEC_2.fasta"

# Create lists to store data
names = []
sequences = []
lengths = []

# Parse the FASTA file
for record in SeqIO.parse(fasta_file, "fasta"):
    names.append(record.description)
    sequences.append(str(record.seq))
    lengths.append(len(record.seq))

# Create a pandas DataFrame
data = {'Name': names, 'Sequence_protein': sequences, 'Sequence Length': lengths}
df = pd.DataFrame(data)

# Display the DataFrame
display(df.head())
```

```python
df_IDR[' FASTA header'] = df_IDR[' FASTA header'].str.lstrip()
```

```python
df['Name'] = df['Name'].str.replace(',', ';')

print("\nDataFrame after replacing ',' with ';':")
display(df.head())
```

```python
merged_df = pd.merge(df_IDR, df, left_on=' FASTA header', right_on='Name', how='left')

merged_df = merged_df.drop('Name', axis=1)

# Display the updated DataFrame
display(merged_df.head())
```

```python
# Define the conditions
conditions = [
    # First condition: 'IDR start' between 1 and 20 (inclusive) and 'IDR end' less than 'sequence length'
    (merged_df[' IDR start'] >= 1) & (merged_df[' IDR start'] <= 20) & (merged_df[' IDR end'] < merged_df['Sequence Length']),
    # Second condition: 'IDR start' is 1 and 'IDR end' is equal to 'sequence length'
    (merged_df[' IDR start'] == 1) & (merged_df[' IDR end'] == merged_df['Sequence Length']),
    # Third condition: 'IDR start' is greater than 1 and 'IDR end' is between 'sequence length' and 20 before (inclusive)
    (merged_df[' IDR start'] > 1) & (merged_df[' IDR end'] <= merged_df['Sequence Length']) & (merged_df[' IDR end'] >= merged_df['Sequence Length'] - 20),
    # Fourth condition: 'IDR start' is greater than 1 and 'IDR end' is less than 'sequence length' (and not covered by the third condition)
    #(merged_df[' IDR start'] > 1) & (merged_df[' IDR end'] < merged_df['Sequence Length']) & (merged_df[' IDR end'] < merged_df['Sequence Length'] - 20)
    (merged_df[' IDR start'] > 1) & (merged_df[' IDR end'] < merged_df['Sequence Length']) &
    ~((merged_df[' IDR start'] >= 1) & (merged_df[' IDR start'] <= 20)) & # Exclude first condition's start range
    ~((merged_df[' IDR end'] <= merged_df['Sequence Length']) & (merged_df[' IDR end'] >= merged_df['Sequence Length'] - 20)) # Exclude third condition's end range
]

# Define the corresponding values for the new 'location' column
values = ['N-terminal', 'IDP', 'C-terminal', 'Middle']

# Create the new 'location' column using np.select
merged_df['location'] = np.select(conditions, values, default='unknown') # 'unknown' for cases that don't match any condition

print("\nDataFrame with 'location' column:")
display(merged_df.head())
```

```python
# Filter the DataFrame for rows where 'location' is 'unknown'
unknown_location_df = merged_df[merged_df['location'] == 'IDP']

print("\nDataFrame with only 'unknown' locations:")
display(unknown_location_df)
```
