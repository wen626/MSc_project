```python
from Bio.Seq import Seq
from Bio.SeqRecord import SeqRecord
from Bio import SeqIO
```

```python
records = []
for index, row in df.iterrows():
    if  len(str(row['Protein_sequence'])) > 0 and len(str(row['Effector'])) > 0 :
        #print(len(str(row['Protein_sequence'])))
        try:
            sequence_id = str(row['Effector'])
            protein_sequence = str(row['Protein_sequence'])
    
            # Create a SeqRecord object for each protein sequence
            record = SeqRecord(
                Seq(protein_sequence),
                id=sequence_id
                #description="" # You can add a description if available in another column
            )
            records.append(record)
        except KeyError as e:
            print(f"Error: Missing column '{e}' in the DataFrame. Please check column names.")
            #return
        except Exception as e:
            print(f"Error processing row {index}: {e}")
            #continue
    else:
        #print(str(row['Protein_sequence']) + str(row['Effector']))
        print(str(row['Protein_sequence']))
        
```

```python
try:
    with open('/home/wenyuantong/Desktop/data/PsyTEC_2.fasta', "w") as output_handle:
        SeqIO.write(records, output_handle, "fasta")
    print(f"FASTA file successfully created at {'/home/wenyuantong/Desktop/data/PsyTEC_2.fasta'}")
except Exception as e:
    print(f"Error writing FASTA file: {e}")
```
