```python
from Bio.Seq import Seq
from Bio.SeqRecord import SeqRecord
from Bio import SeqIO
```

```python
records = []
for index, row in filtered_df.iterrows():
    try:
        sequence_id = str(row['IDR ID'])
        protein_sequence = str(row[' sequence'])
        #type_protein =str(row['FinalType'])

        # Create a SeqRecord object for each protein sequence
        record = SeqRecord(
            Seq(protein_sequence),
            id=sequence_id,
            description=''# You can add a description if available in another column
        )
        records.append(record)
    except KeyError as e:
        print(f"Error: Missing column '{e}' in the DataFrame. Please check column names.")
        #return
    except Exception as e:
        print(f"Error processing row {index}: {e}")
        #continue
```

```python
print(records[88])
```

```python
try:
    with open('/home/wenyuantong/Desktop/data/filtered_IDR_Arabi.fasta', "w") as output_handle:
        SeqIO.write(records, output_handle, "fasta")
    print(f"FASTA file successfully created at {'/home/wenyuantong/Desktop/data/filtered_IDR_Arabi.fasta'}")
except Exception as e:
    print(f"Error writing FASTA file: {e}")
```
