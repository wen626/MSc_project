```python
import pandas as pd


file_path = 'RIN4_domain.tsv'

try:
    df_rin4 = pd.read_csv(file_path, sep='\t', header = None)
    print("RIN4_domain.tsv loaded successfully!")
    display(df_rin4.head())
except FileNotFoundError:
    print(f"Error: The file '{file_path}' was not found. Please check the path and try again.")
except Exception as e:
    print(f"An error occurred while reading the file: {e}")
```

```python
display(df_rin4.tail())
```

```python
# Define the minimum and maximum allowed lengths for segments
MIN_LENGTH = 30
MAX_LENGTH = 80

# List to store the new, processed rows
processed_rows = []

# Iterate through each row of the original DataFrame
for index, row in df_rin4.iterrows():
    protein_name = row[0]
    start_pos = row[1]
    end_pos = row[2]
    domain_type = row[3]

    # Calculate the current length of the entry
    current_length = end_pos - start_pos + 1

    # If the entry is too long, split it into segments
    if current_length > MAX_LENGTH:
        segment_start = start_pos
        segment_count = 1
        while segment_start <= end_pos:
            # Determine the end position for the current segment, ensuring it doesn't exceed MAX_LENGTH or the original end_pos
            segment_end = min(segment_start + MAX_LENGTH - 1, end_pos)
            segment_length = segment_end - segment_start + 1

            # Only add the segment if its length is within the desired range
            if segment_length >= MIN_LENGTH:
                processed_rows.append([protein_name, segment_start, segment_end, f"{domain_type}_{segment_count}"])

            # Move to the start of the next segment
            segment_start = segment_end + 1
            segment_count += 1
    # If the entry is within the allowed length range, add it as is (with a '_1' suffix as per example)
    elif current_length >= MIN_LENGTH and current_length <= MAX_LENGTH:
        processed_rows.append([protein_name, start_pos, end_pos, f"{domain_type}_1"])
    # Entries shorter than MIN_LENGTH are implicitly dropped due to the conditions above

# Create the new DataFrame from the processed rows
df_processed = pd.DataFrame(processed_rows, columns=df_rin4.columns)

print("New DataFrame with entries segmented to be between 30 and 80 in length:")
display(df_processed)
```

```python
display(df_processed.tail())
```

```python
#The new DataFrame `df_processed` now contains entries that are all within the 30 to 80 length range
#Longer original entries have been split, and their `Domain_Type` column has been suffixed with `_1`, `_2`, etc
#Entries shorter than 30 have been excluded.
```

```python
output_file_path = 'IDR_arabi_effector_domain_cut.tsv'

try:
    df_processed.to_csv(output_file_path, sep='\t', index=False)
    print(f"DataFrame successfully saved to '{output_file_path}'")
except Exception as e:
    print(f"An error occurred while saving the file: {e}")
```
